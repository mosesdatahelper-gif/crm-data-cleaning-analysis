```python
lead_source_analysis = df.groupby("Lead_Source").agg(
    Leads=("Customer_ID", "count"),
    Total_Deal_Value=("Deal_Value", "sum"),
    Average_Deal_Value=("Deal_Value", "mean"),
    Won=("Lead_Status", lambda x: (x == "Won").sum())
)

lead_source_analysis["Win_Rate"] = (
    lead_source_analysis["Won"] / lead_source_analysis["Leads"] * 100
)

lead_source_analysis = lead_source_analysis.sort_values(
    "Total_Deal_Value", ascending=False
)

lead_source_analysis
```

```python
industry_analysis = df.groupby("Industry").agg(
    Leads=("Customer_ID", "count"),
    Total_Deal_Value=("Deal_Value", "sum"),
    Average_Deal_Value=("Deal_Value", "mean"),
    Won=("Lead_Status", lambda x: (x == "Won").sum())
)

industry_analysis["Win_Rate"] = (
    industry_analysis["Won"] / industry_analysis["Leads"] * 100
)

industry_analysis
```

```python
country_analysis = df.groupby("Country").agg(
    Leads=("Customer_ID", "count"),
    Total_Deal_Value=("Deal_Value", "sum"),
    Average_Deal_Value=("Deal_Value", "mean"),
    Won=("Lead_Status", lambda x: (x == "Won").sum())
)

country_analysis["Win_Rate"] = (
    country_analysis["Won"] / country_analysis["Leads"] * 100
)

country_analysis
```
